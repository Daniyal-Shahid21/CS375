## <font color='blue'> Question 0. MLP with scikit-learn: Explore outputs for the MNIST dataset </font>

The purpose of this question is to get some practice time with MLP classifiers. We will be using the the MNIST data set.




```python
# Let's first load the MNIST data set.

from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split


X, y = fetch_openml('mnist_784', version=1, return_X_y=True)
y = y.astype(int)
X = ((X / 255.) - .5) * 2
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=10000, random_state=123, stratify=y)
```

<br> **Q0-0**: This the classifier we built in class. Find a setting for *batch_size* and *max_iter*, so that the subsequent fit function **converges**.





(If for this question you have used LLMs such as ChatGPT, replace this text by screenshots of the full transcript of your session. Create extra text cells if needed.)

```python

mlp_mnist = MLPClassifier(verbose = True, hidden_layer_sizes=(50,), batch_size = 256, \
                          max_iter=200, solver='sgd', activation='logistic',\
                          learning_rate = 'constant', learning_rate_init = 0.001, random_state=1)

#batch_size and max_iter were both changed

```

```python
mlp_mnist.fit(X_train,y_train)
print("n_iter_ =", mlp_mnist.n_iter_)
print("loss_   =", mlp_mnist.loss_)
```

```python
predictions = mlp_mnist.predict(X_test)
probability_predictions = mlp_mnist.predict_proba(X_test)

print(predictions)
print(probability_predictions)
```

**Q0-1**: Write a short function that outputs the **second best** prediction for each point &mdash; that is, the label that gets the second highest probability in the softmax output.

(If for this question you have used LLMs such as ChatGPT, replace this text by screenshots of the full transcript of your session. Create extra text cells if needed.)

```python
def second_best_labels(probability_predictions):
    return np.argsort(probability_predictions, axis=1)[:, -2]

second_best = second_best_labels(probability_predictions)

print(second_best)
```

**Q0-2**: For each data point misclassified by the model, determine whether the second best prediction is actually the correct label. Calculate the percentage of those misclassified points that are correctly classified based on the second best prediction.

(If for this question you have used LLMs such as ChatGPT, replace this text by screenshots of the full transcript of your session. Create extra text cells if needed.)

```python
# predicted labels
pred = predictions

# true labels
true = y_test.to_numpy() if hasattr(y_test, "to_numpy") else np.array(y_test)

# identify misclassified values
misclassified = (pred != true)

# check if second best prediction equals the true label
second_best_correct = (second_best == true)

# percentage of misclassified points fixed by second best predict
percentage = 100 * second_best_correct[misclassified].mean()

print("Percentage of misclassified points corrected by 2nd best predict:", percentage)
print("# of misclassified points:", misclassified.sum())
```

**Q0-3**: From a human point of view, we can perhaps see a frequent similarity between handwritten digits 3 and 5 (sometimes we have to pay closer attention to figure out what the true written digit is). We may want to ask if the MLP also detects digit similarities.    
<br>
Consider then the subset $X_i$ of data points that have label $i$ (e.g. all data points with label 5). For each label $j\neq i$, report the frequency by which label $j$ (e.g. label 3) shows up as the second best prediction for points in $X_i$.
<br>
<br>
***Further Clarification***: Suppose we have 100 data points with label i=5. Let's say that for 28 of these 100 points, the second predicted label is j=3. Then, the frequency of j=3 as second-best prediction to i=5 is 28%.  We want to compute that frequency for every i and j.
<br>  
Give your code and comment on your findings.

(If for this question you have used LLMs such as ChatGPT, replace this text by screenshots of the full transcript of your session. Create extra text cells if needed.)

```python
freq = np.zeros((10, 10), dtype=float)

for i in range(10):
    idx = (true == i)                 # points with true label i
    sb = second_best[idx]             # 2nd best predicts for those points
    counts = np.bincount(sb, minlength=10)
    freq[i] = counts / idx.sum()      # convert to frequencies

# print results as %, excluding diagonal
for i in range(10):
    print(f"\nTrue digit i={i}: second-best distribution (%)")
    for j in range(10):
        if j != i:
            print(f"  j={j}: {100*freq[i,j]:5.1f}%")
```

**Q1-0**. MLP classifiers in scikit-learn have an option for so-called 'warm start'. Do a bit of research to find out what it is. Then:

Give an example of warm start. Specifically
train the network for $k$ epochs, and then using warm start
continue the training for another $k$ epochs (4 points).
Here $k$ is a parameter of your choice.  





(If for this question you have used LLMs such as ChatGPT, replace this text by screenshots of the full transcript of your session. Create extra text cells if needed.)

```python
from sklearn.neural_network import MLPClassifier

k = 50  # epochs per stage

mlp_ws = MLPClassifier(
    verbose=True,
    hidden_layer_sizes=(50,),
    solver="sgd",
    activation="logistic",
    learning_rate="constant",
    learning_rate_init=0.001,
    batch_size=256,
    max_iter=k,          # train k epochs
    warm_start=True,     # keep weights for next fit()
    random_state=1
)

# k epochs
mlp_ws.fit(X_train, y_train)
loss_1 = mlp_ws.loss_
iters_1 = mlp_ws.n_iter_

# Stage 2: continue training for more k epochs
mlp_ws.set_params(max_iter=k)   # keep k epochs for this second fit call
mlp_ws.fit(X_train, y_train)
loss_2 = mlp_ws.loss_
iters_2 = mlp_ws.n_iter_

print("Stage 1 epochs run:", iters_1, "loss:", loss_1)
print("Stage 2 epochs run:", iters_2, "loss:", loss_2)
```

**Q2-1.** With the label of the image 'planted' in the image itself, a Neural Network can achieve accuracy 100%. This is because all the weights connecting to pixels $2...28^2$ can be set to 0, and so effectively the only feature that will be taken into account is the first pixel. The question is whether an MLP can indeed 'discover' this weighting via training.
<br>
<br>
Use an MLP classifier  with one hidden layer of any size you want. Train it on $X'$ and report the accuracy. Does it reach 100\%? Is the accuracy better relative to training with the original input $X$?


Yes, an MLP can discover this weighting via training.

Because the true label is directly encoded in the first pixel, the learning problem becomes trivial: there exists a solution where the network assigns a strong weight to the first input feature and drives the weights of all remaining pixels toward zero. During training, gradient descent will naturally favor this solution because it minimizes the loss most efficiently while changing the weight on the first pixel alone is sufficient to perfectly predict the label, the other pixels add noise rather than useful signal.

You should observe near 100% training and test accuracy. However, this can also indicate that the network has “cheated” by learning to read the planted label from the first pixel instead of learning visual digit structure.
